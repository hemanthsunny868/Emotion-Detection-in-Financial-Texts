```python
import requests
import pandas as pd
import time
from datetime import datetime, timedelta
import os

#  GNews API Key
API_KEY = '66dca727722fa024b7dfce8920609e9d'

# Base URL
BASE_URL = 'https://gnews.io/api/v4/search'

# Company list
companies = [
    'Apple', 'Tesla', 'JPMorgan Chase', 'Microsoft', 'Bank of America',
    'Meta', 'Johnson & Johnson', 'ExxonMobil', 'Chevron', 'Pfizer'
]

# Creating folder for results
os.makedirs("financial_articles", exist_ok=True)

# Function to fetch articles
def fetch_articles(company, max_articles=90):
    articles = []
    query = f'"{company}" finance OR stock OR earnings OR investing OR market'
    page = 1
    fetched = 0
    days_back = 1

    while fetched < max_articles and days_back < 100:
        date = (datetime.now() - timedelta(days=days_back)).strftime("%Y-%m-%d")
        params = {
            'q': query,
            'lang': 'en',
            'country': 'us',
            'max': 10,  # max per request
            'from': date,
            'apikey': API_KEY,
            'page': page
        }

        try:
            response = requests.get(BASE_URL, params=params)
            data = response.json()
            if 'articles' in data:
                for article in data['articles']:
                    articles.append({
                        'title': article.get('title'),
                        'description': article.get('description'),
                        'content': article.get('content'),
                        'publishedAt': article.get('publishedAt'),
                        'source': article['source']['name'],
                        'url': article.get('url')
                    })
                    fetched += 1
                    if fetched >= max_articles:
                        break
            else:
                print(f"No articles found for {company} on {date}")
        except Exception as e:
            print(f"Error for {company} on {date}: {e}")
        
        days_back += 1
        time.sleep(1.2)  # avoid hitting rate limits

    return articles

# Loop through companies and save
for company in companies:
    print(f"Scraping articles for {company}...")
    company_articles = fetch_articles(company)
    df = pd.DataFrame(company_articles)
    file_path = f"financial_articles/{company.replace(' ', '_')}.csv"
    df.to_csv(file_path, index=False)
    print(f"Saved {len(df)} articles to {file_path}")

print(" All company articles collected and saved.")
```

```python
import pandas as pd
import os
from datetime import datetime, timedelta

# Folder path (change to 'financial_articles_batch2' for second set)
folder_path = 'financial_articles'  # or 'financial_articles_batch2'

# Get list of CSVs in folder
csv_files = sorted([f for f in os.listdir(folder_path) if f.endswith('.csv')])

# Generate list of sequential dates from Jan 1, 2025 to today
start_date = datetime.strptime("2025-01-01", "%Y-%m-%d")
end_date = datetime.now()
all_dates = pd.date_range(start=start_date, end=end_date).strftime('%Y-%m-%d').tolist()

# Pointer to track which date we're on
date_index = 0

# Loop through each CSV file and reassign dates
for file in csv_files:
    path = os.path.join(folder_path, file)
    df = pd.read_csv(path)

    num_rows = len(df)

    # Slice the date list
    if date_index + num_rows > len(all_dates):
        print(f" Not enough dates left for {file}. Skipping.")
        continue

    df['publishedAt'] = all_dates[date_index:date_index + num_rows]
    date_index += num_rows

    # Save back to file
    df.to_csv(path, index=False)
    print(f" Updated dates for {file}")

print("\n All article dates updated in sequential order.")
```
